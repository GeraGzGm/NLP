```python
import numpy as np
import string
```

```python
class Cypher:
    random_order = list(string.ascii_lowercase)
    np.random.shuffle(random_order)
    encoder = {char:new_val for char, new_val in zip(list(string.ascii_lowercase), random_order)}


    @classmethod
    def encode_message(cls, message: str):
        return ''.join(cls.encoder.get(char, char) for char in message.lower())
        
```

```python
Cypher.encode_message("hola perro")
```

```python
from nltk.corpus import brown
import nltk
from collections import Counter
import numpy as np

nltk.data.path.append("../datasets/nltk")
nltk.download("brown", download_dir = "../datasets/nltk")

# Prepare bigrams
bigrams = Counter()
for sentence in brown.sents():
    words = ' '.join(sentence).lower()
    for i in range(len(words)-1):
        bigrams[words[i:i+2]] += 1

# Normalize to probabilities
total = sum(bigrams.values())
bigram_probs = {k: np.log(v/total) for k, v in bigrams.items()}
```
